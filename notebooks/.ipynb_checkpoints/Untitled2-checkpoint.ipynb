```python
#importing libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
import joblib
import os
```

```python
#Load Processed Data for One Stock (Tesla first)
df = pd.read_csv("../data/processed/TSLA.csv", index_col="Date")
df.head()
```

```python
#Define Features & Targets
X=df.drop("Target", axis=1)
y=df["Target"]
```

```python
X.shape, y.shape
```

```python
#Train/Test Split
train_size = int(len(X) * 0.8)

X_train = X.iloc[:train_size]
X_test = X.iloc[train_size:]

y_train = y.iloc[:train_size]
y_test = y.iloc[train_size:]

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
#Train Random Forest Model
model = RandomForestClassifier(
    n_estimators=400,
    max_depth=8,
    min_samples_split=10,
    min_samples_leaf=5,
    max_features="sqrt",
    class_weight="balanced",
    n_jobs=-1,
    random_state=42
)

model.fit(X_train, y_train)
```

```python
#Predictions
y_pred=model.predict(X_test)
```

```python
#Evaluate Model
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

```python
#Confusion Matrix
print(confusion_matrix(y_test, y_pred))
```

```python
#Classification Report
print(classification_report(y_test, y_pred))
```

```python
#Save Model
os.makedirs("../models", exist_ok=True)
joblib.dump(model, "../models/TSLA_rf.pkl")
```

```python
import pandas as pd
import joblib
import os

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

from xgboost import XGBClassifier


tickers = ["TSLA", "AMZN", "NVDA", "MSFT", "GOOG"]

FEATURES = [
    'MA20', 'MA50',
    'RSI',
    'MACD', 'MACD_signal', 'MACD_hist',
    'BB_width',
    'Return',
    'Momentum3', 'Momentum10',
    'Trend',
    'Volatility10',
    'Lag1', 'Lag5'
]


for ticker in tickers:
    print(f"\n======= Training {ticker} =======")

    # Load processed data
    df = pd.read_csv(f"../data/processed/{ticker}.csv", index_col=0)

    # Ensure only desired features
    df = df[FEATURES + ['Target']]

    # Drop NaN
    df.dropna(inplace=True)

    # Split
    X = df[FEATURES]
    y = df['Target']

    train_size = int(len(X) * 0.8)
    X_train, X_test = X.iloc[:train_size], X.iloc[train_size:]
    y_train, y_test = y.iloc[:train_size], y.iloc[train_size:]

    # Scaling
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    # ========================================
    # Logistic Regression
    # ========================================

    log_model = LogisticRegression(max_iter=1000)
    log_model.fit(X_train_scaled, y_train)
    log_preds = log_model.predict(X_test_scaled)
    log_acc = accuracy_score(y_test, log_preds)
    print(f"Logistic Accuracy: {log_acc:.4f}")

    # ========================================
    # Random Forest
    # ========================================

    rf_model = RandomForestClassifier(
        n_estimators=400,
        max_depth=6,
        min_samples_split=20,
        min_samples_leaf=10,
        max_features='sqrt',
        class_weight='balanced',
        n_jobs=-1,
        random_state=42
    )

    rf_model.fit(X_train_scaled, y_train)
    rf_preds = rf_model.predict(X_test_scaled)
    rf_acc = accuracy_score(y_test, rf_preds)
    print(f"RandomForest Accuracy: {rf_acc:.4f}")

    # ========================================
    # XGBOOST
    # ========================================

    xgb_model = XGBClassifier(
        n_estimators=300,
        max_depth=4,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_lambda=1.0,
        reg_alpha=0.0,
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1,
        tree_method="hist"  
    )

    xgb_model.fit(X_train_scaled, y_train)
    xgb_preds = xgb_model.predict(X_test_scaled)
    xgb_acc = accuracy_score(y_test, xgb_preds)
    print(f"XGBoost Accuracy: {xgb_acc:.4f}")

    # ========================================
    # Select Best Model
    # ========================================

    models = {
        "LogisticRegression": (log_acc, log_model),
        "RandomForest": (rf_acc, rf_model),
        "XGBoost": (xgb_acc, xgb_model)
    }

    best_name = max(models, key=lambda m: models[m][0])
    best_acc, best_model = models[best_name]

    print(f"\n Best Model: {best_name} ({best_acc:.4f})")

    # Save
    os.makedirs("../models", exist_ok=True)
    joblib.dump(best_model, f"../models/{ticker}_model.pkl")
    joblib.dump(scaler, f"../models/{ticker}_scaler.pkl")

    print(f"Saved: {ticker}_model.pkl")
    print(f"Saved: {ticker}_scaler.pkl")

    # Report
    preds = best_model.predict(X_test_scaled)
    


    print("\nReport:")
    print(classification_report(y_test, preds, digits=4))
    
```

Feature Importance Visualization

```python

def plot_feature_importance(model, features, ticker, model_name):
    import numpy as np
    import pandas as pd
    import matplotlib.pyplot as plt
    import seaborn as sns

    # Tree-based models
    if hasattr(model, "feature_importances_"):
        importances = model.feature_importances_

    # Linear models
    elif hasattr(model, "coef_"):
        importances = model.coef_[0]

    else:
        print(f"No feature importance for model: {model_name}")
        return
    
    fi = pd.DataFrame({"feature": features, "importance": importances})
    fi = fi.sort_values("importance", key=np.abs, ascending=True)

    plt.figure(figsize=(8,6))
    sns.barplot(data=fi, x="importance", y="feature")
    plt.axvline(0, color='red', linestyle='--')
    plt.title(f"{ticker} Feature Importance ({model_name})")
    plt.tight_layout()
    plt.show()
```

```python
plot_feature_importance(best_model, FEATURES, ticker, best_name)
```

ROC Curve / AUC

```python
from sklearn.metrics import roc_curve, auc

probs = best_model.predict_proba(X_test_scaled)[:, 1]
fpr, tpr, thresh = roc_curve(y_test, probs)
auc_score = auc(fpr, tpr)

plt.figure(figsize=(6,6))

plt.plot(fpr, tpr, label=f"ROC Curve (AUC = {auc_score:.2f})")
plt.plot([0, 1], [0, 1], '--', label="Random", alpha=0.7)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title(f"{ticker} ROC Curve (AUC={auc_score:.2f})")
plt.legend()
plt.grid(True)
plt.show()
```
